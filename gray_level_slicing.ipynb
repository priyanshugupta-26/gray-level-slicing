```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read coin image
image_path = "/content/coin.jpeg"
img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

if img is None:
    print("Image not found! Check the path.")
else:
    rows, cols = img.shape

    # Create output images
    without_bg = np.zeros((rows, cols), dtype=np.uint8)
    with_bg = img.copy()

    # Gray level range
    min_range = 100
    max_range = 150

    # Apply Gray Level Slicing
    for i in range(rows):
        for j in range(cols):
            pixel = img[i, j]

            if min_range <= pixel <= max_range:
                without_bg[i, j] = 255
                with_bg[i, j] = 255
            else:
                without_bg[i, j] = 0
                with_bg[i, j] = pixel

    # Display output
    plt.figure(figsize=(12, 4))

    plt.subplot(1, 3, 1)
    plt.imshow(img, cmap="gray", vmin=0, vmax=255)
    plt.title("Original Coin Image")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(without_bg, cmap="gray", vmin=0, vmax=255)
    plt.title("Slicing Without Background")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(with_bg, cmap="gray", vmin=0, vmax=255)
    plt.title("Slicing With Background")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```
